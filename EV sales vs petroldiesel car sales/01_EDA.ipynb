## Box plot Inference 
I used boxplots to identify the spread and potential outliers in the numerical variables. The analysis showed significant variation and extreme values in several sales and infrastructure variables. These outliers represent large differences between countries and years, so they should be investigated before machine learning rather than removed blindly.

A boxplot is mainly used to identify:

Median — central line inside the box
Q1 and Q3 — lower and upper quartiles
IQR — spread of the middle 50% of values
Whiskers — typical range of observations
Outliers — points outside the whiskers


```python
for column in numerical_columns:

    plt.figure(figsize=(7, 4))

    sns.boxplot(
        x=df[column]
    )

    plt.title("Boxplot of " + column)

    plt.show()
```

## CATEGORICAL ANALYSIS inference
I used countplots to understand the distribution of categorical variables. The analysis confirmed that the dataset covers 25 countries, 5 regions and 3 vehicle segments. The powertrain variable is highly imbalanced because ICE records are much more common than BEV records. This imbalance needs to be considered during the machine-learning stage.

Variable-wise inference :

country -

## EV VS PETROL SALES Inference
I calculated the total EV and petrol sales to compare their overall sales volume. The analysis shows that petrol sales are much higher than EV sales across the dataset, indicating that petrol vehicles still have a larger overall market presence. However, EV adoption should also be analysed using EV market share and yearly trends

```python
total_ev = df["ev_sales"].sum()

total_petrol = df["petrol_car_sales"].sum()

print("Total EV Sales:", total_ev)
print("Total Petrol Sales:", total_petrol)
```

I created a bar chart to visually compare the total EV and petrol sales. The chart clearly shows that petrol sales are much higher than EV sales, indicating that petrol vehicles still dominate the overall sales volume in the dataset. However, EV adoption should also be evaluated using market share and yearly trends

```python
sales = {
    "EV": total_ev,
    "Petrol": total_petrol
}

plt.figure(figsize=(7, 5))

plt.bar(
    sales.keys(),
    sales.values()
)

plt.title("EV vs Petrol Sales")

plt.xlabel("Vehicle Type")

plt.ylabel("Sales")

plt.show()
```

## EV Sales Trend Inference
The EV sales trend shows an overall increasing pattern in EV sales over the 2010–2025 period. This indicates that EV adoption has expanded significantly over time. The upward movement reflects the gradual transition from conventional vehicles toward electric vehicles. However, the year-to-year growth is not necessarily uniform, and some periods may show sharper increases than others.

```python
year_ev = df.groupby(
    "year"
)["ev_sales"].sum()

plt.figure(figsize=(10, 5))

plt.plot(
    year_ev.index,
    year_ev.values,
    marker="o"
)

plt.title("EV Sales Trend")

plt.xlabel("Year")

plt.ylabel("EV Sales")

plt.show()
```

## Petrol Car Sales Trend Inference
The petrol car sales trend shows the year-by-year movement of conventional petrol vehicle sales from 2010 to 2025. Compared with the EV trend, petrol vehicles continue to represent a much larger overall sales volume in the dataset. The trend can be used to identify whether petrol sales are increasing, declining, or fluctuating over the study period and to compare the transition between conventional and electric vehicles.

The overall dataset contains 927,967,539 petrol car sales, compared with 76,322,837 EV sales, confirming the much larger aggregate petrol-sales volume.

```python
year_petrol = df.groupby(
    "year"
)["petrol_car_sales"].sum()

plt.figure(figsize=(10, 5))

plt.plot(
    year_petrol.index,
    year_petrol.values,
    marker="o"
)

plt.title("Petrol Car Sales Trend")

plt.xlabel("Year")

plt.ylabel("Petrol Sales")

plt.show()
```

## EV Sales by Region Inference 
The analysis shows that EV sales are not evenly distributed across regions. Some regions contribute substantially more to total EV sales than others. This indicates that EV adoption and EV market size vary geographically. The regional differences may be related to factors such as market size, charging infrastructure, economic conditions, fuel prices, EV incentives, and environmental policies, which can be investigated in subsequent EDA.

Important point:


The number of records in each region is not equal in your dataset, so you should be careful when interpreting total EV sales by region. A region having more records can naturally have a larger aggregate.

For example, your dataset has Europe as the largest region by record count, so total regional sales should ideally be complemented with EV market share for a fairer adoption comparison.


```python
region_ev = df.groupby(
    "region"
)["ev_sales"].sum().sort_values(
    ascending=False
)

plt.figure(figsize=(10, 5))

region_ev.plot(kind="bar")

plt.title("EV Sales by Region")

plt.xlabel("Region")

plt.ylabel("EV Sales")

plt.xticks(rotation=45)

plt.tight_layout()

plt.show()
```

## Sub plot
The four distributions are highly uneven, with EV sales, petrol sales, EV market share, and charging infrastructure showing strong right-skewness. Most observations represent relatively low values, while a smaller number of country-year observations have extremely high sales, market penetration, or infrastructure levels. This confirms the presence of substantial variation and extreme observations, which should be considered during feature engineering and machine learning.

```python
fig, axes = plt.subplots(
    2,
    2,
    figsize=(14, 10)
)


sns.histplot(
    df["ev_sales"],
    kde=True,
    ax=axes[0, 0]
)

axes[0, 0].set_title(
    "EV Sales Distribution"
)



sns.histplot(
    df["petrol_car_sales"],
    kde=True,
    ax=axes[0, 1]
)

axes[0, 1].set_title(
    "Petrol Sales Distribution"
)



sns.histplot(
    df["ev_market_share"],
    kde=True,
    ax=axes[1, 0]
)

axes[1, 0].set_title(
    "EV Market Share Distribution"
)



sns.histplot(
    df["charging_stations"],
    kde=True,
    ax=axes[1, 1]
)

axes[1, 1].set_title(
    "Charging Stations Distribution"
)

plt.tight_layout()

plt.show()
```
